3.1.

 Берілген уақыттық қатар:

 [2, 4, 6, 8, 10]

 RNN көмегімен келесі мәнді болжау идеясының логикасын түсіндір.

```python
import numpy as np
```

```python
def simple_rnn(sequence):
    h = 0

    for x in sequence:
        h = h + (x - h) * 0.5  # Өткенді есте сақтау
        print(f"Кіріс: {x}, Жасырын күй: {h:.1f}")

    difference = sequence[-1] - sequence[-2]
    prediction = sequence[-1] + difference

    return prediction
```

```python
sequence = [2, 4, 6, 8, 10]
result = simple_rnn(sequence)

print(f"Болжам: {result}")
```


3.2.

 Мәтіндік деректер үшін RNN қолданғанда:

input қалай беріледі?

one-hot encoding не үшін керек?


________________________________

1. Input Қалай Беріледі?

Мысал мәтін:

"hello"
Қадам 1: Токенизация (Сөздерге бөлу)

    ['h', 'e', 'l', 'l', 'o']

Қадам 2: Сөздікке айналдыру

Сөздік:

    'h' → 0

    'e' → 1

    'l' → 2

    'o' → 3

Қадам 3: Сандарға түрлендіру

"hello" → [0, 1, 2, 2, 3]

Қадам 4: RNN-ге беру

     x₁=0    x₂=1    x₃=2    x₄=2    x₅=3
      ↓       ↓       ↓       ↓       ↓
    [RNN] → [RNN] → [RNN] → [RNN] → [RNN]

Әр уақыт қадамында: бір әріп (немесе сөз) беріледі

_________________

Неге One-Hot Керек?

1. Математикалық қатынасты алып тастайды

        Жаман:  'a'=0, 'z'=25  →  'z' > 'a' ❌

        Жақсы:  'a'=[1,0,...], 'z'=[0,...,1]  →  теңдей ✅

2. Нейрондық желі түсінеді

        One-hot вектор = желі үшін түсінікті формат

3. Әр әріп бірегей

        'h' ≠ 'e' ≠ 'l' ≠ 'o'  (барлығы ерекше)

```python
import numpy as np


text = "hello"


chars = sorted(set(text))
char_to_idx = {ch: i for i, ch in enumerate(chars)}

print("Сөздік:", char_to_idx)
# {'e': 0, 'h': 1, 'l': 2, 'o': 3}

#  Сандарға түрлендіру
indices = [char_to_idx[ch] for ch in text]
print("Индекстер:", indices)
# [1, 0, 2, 2, 3]

# One-hot encoding
vocab_size = len(chars)
one_hot = np.zeros((len(text), vocab_size))

for i, idx in enumerate(indices):
    one_hot[i, idx] = 1

print("\nOne-hot vectors:")
print(one_hot)
```

--------------------------

    2023 жаз (↓) → 2023 күз (↑) → 2024 қыс (↑↑) → 2024 көктем (?)
        ↓              ↓              ↓              ↓

3. Медициналық Диагностика (Medical Diagnosis)

Мәселе:

Науқастың бүкіл медициналық тарихы маңызды:

- 5 жыл бұрын: жүрек ауруы
- 2 жыл бұрын: диабет
- Қазір: жоғары қан қысымы

Диагноз: барлық тарихты ескеру керек!

Неге LSTM керек:

    Ұзақ мерзімді науқас тарихы: 10 жыл бұрынғы науқаслық маңызды
    Себеп-салдар байланысы: бұрынғы аурулар қазіргіге әсер етеді
    Қарапайым RNN: тек соңғы 3-5 визитті есте сақтайды
    LSTM: толық медициналық профильді сақтайды

____________


    2018: операция → 2020: аллергия → 2023: жүрек ауруы → 2025: ???
            ↓              ↓                ↓               ↓
        LSTM барлық тарихты талдап, дұрыс диагноз қояды

```python
import numpy as np
from keras.models import Sequential
from keras.layers import SimpleRNN, Dense
from keras.utils import to_categorical
```

```python
# 1. ДЕРЕК ДАЙЫНДАУ
text = "hello"
chars = sorted(set(text))
char_to_idx = {ch: i for i, ch in enumerate(chars)}
idx_to_char = {i: ch for ch, i in char_to_idx.items()}

vocab_size = len(chars)
print(f"Сөздік: {char_to_idx}")
print(f"Өлшем: {vocab_size}")
```

```python
# Input: "hell" → Output: "o"
sequence = [char_to_idx[ch] for ch in text]
X = []
y = []
```

```python
# Әр 4 әріптен келесі әріпті болжаймыз
seq_length = 4
for i in range(len(sequence) - seq_length):
    X.append(sequence[i:i+seq_length])
    y.append(sequence[i+seq_length])

X = np.array(X)
y = np.array(y)

print(f"X shape: {X.shape}")
print(f"y shape: {y.shape}")
print(f"X: {X}")  # [[0 1 2 2]] → "hell"
print(f"y: {y}")  # [3] → "o"
```

```python
# ONE-HOT ENCODING
X = to_categorical(X, num_classes=vocab_size)
y = to_categorical(y, num_classes=vocab_size)

print(f"X one-hot shape: {X.shape}")
print(f"y one-hot shape: {y.shape}")
```

```python
model = Sequential([
    SimpleRNN(units=50,
              activation='relu',
              input_shape=(seq_length, vocab_size)),  # (4, 4)
    Dense(vocab_size, activation='softmax')
])
```

```python
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
model.summary()
```

```python
history = model.fit(
    X, y,
    epochs=500,
    verbose=0  # Тек соңғы нәтижені көрсет
)

print(f"Соңғы accuracy: {history.history['accuracy'][-1]:.2%}")
print(f"Соңғы loss: {history.history['loss'][-1]:.4f}")
```

```python
test_input = X[0].reshape(1, seq_length, vocab_size)
prediction = model.predict(test_input, verbose=0)
predicted_idx = np.argmax(prediction)
predicted_char = idx_to_char[predicted_idx]

print(f"Input: 'hell'")
print(f"Predicted: '{predicted_char}'")
```
