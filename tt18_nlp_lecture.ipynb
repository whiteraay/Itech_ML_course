А деңгейі: Мәтінді тазалау жэне BoW

Темендегі мәтіндер тізімін алып, барлық еріптерді кішірейтіңіз (lowercase), тыныс белгілерін алып тастаңыз жене оларды CountVectorizer көмегімен санға айналдырыңыз.

---

Орнату

    pip install scikit-learn nltk

```python
data = [
    "Алматы - әдемі қала.",
    "Астана - бас қала!",
    "Алматы мен Астана - үлкен қалалар. "
    ]
```

```python
import re
from sklearn.feature_extraction.text import CountVectorizer
```

```python
def clean(text):
    text = text.lower()
    text = re.sub(r'[^\w\s]', '', text) # тыныс белгілерді өшіру
    return text
```

```python
clean_data = [clean(d) for d in data]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(clean_data)

print("Сөздер:", vectorizer.get_feature_names_out())
print("BoW матрица:\n", X.toarray())
```



---


Тапсырма: Осы мәтіндер үшін жиілік матрицасын құрыңыз

В денгейі: TF-IDF қолдану

Sklearn кітапханасын пайдаланып, берілген кұжаттар жиынтығы үшін ең маңызды 3 сөзді
табыңыз (TF-IDF салмағы ең жоғары сөздер)

---



```python
from sklearn.feature_extraction.text import TfidfVectorizer
import numpy as np
```

```python
tfidf = TfidfVectorizer()
X = tfidf.fit_transform(clean_data)

words = tfidf.get_feature_names_out()
scores = X.toarray().sum(axis=0)

top_ids = np.argsort(scores)[-3:]

print("Ең маңызды 3 сөз:")
for i in top_ids:
    print(words[i])
```

transformers

кітапханасын қолданып, келесі екі сөйлемнің, BERT эмбеддингтерін
(векторларын) алаңыз және олардың арасындағы ұқсастықты (similarity) есептеңіз

---




```python
from transformers import AutoTokenizer, AutoModel
import torch
from sklearn.metrics.pairwise import cosine_similarity
```

```python
sentence1 = "I love machine learning."
sentence2 = "AI is amazing."
```

```python
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
model = AutoModel.from_pretrained("bert-base-uncased")
```

```python
def get_embedding(text):
    inputs = tokenizer(text, return_tensors="pt")
    outputs = model(**inputs)
    return outputs.last_hidden_state.mean(dim=1).detach().numpy()
```

```python
emb1 = get_embedding(sentence1)
emb2 = get_embedding(sentence2)

sim = cosine_similarity(emb1, emb2)
print("Ұқсастық:", sim[0][0])
```

Сипаттама: Сізде 100 хат бар (бір бөлігі спам, бір бөлігі пайдалы хаттар).

1. Матіндерді TF-IDF аркылы векторга айналдырыныз.
2. LogisticRegression немесе Naive Bayes классификаторын қолданып, хаттың спам

```python
import pandas as pd
import numpy as np
import re
import nltk


```

```python
df = pd.read_csv("spam.csv", encoding="latin-1")

df = df[['v1','v2']]
df.columns = ['label','text']

print(df.head())
print(df['label'].value_counts())
```

```python
spam_df = df[df['label']=='spam']
ham_df  = df[df['label']=='ham']

spam_sample = spam_df.sample(50, random_state=42)
ham_sample  = ham_df.sample(50, random_state=42)

df = pd.concat([spam_sample, ham_sample])
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

print(df['label'].value_counts())
```

```python
df['label'] = df['label'].map({'ham':0, 'spam':1})
```

```python
df.head()
```

cleaning text

```python
stop_words = set(stopwords.words('english'))

def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    words = text.split()
    words = [w for w in words if w not in stop_words]
    return " ".join(words)

df['clean_text'] = df['text'].apply(clean_text)
```

```python
from nltk.corpus import stopwords
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

nltk.download('stopwords')
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df['clean_text'],
    df['label'],
    test_size=0.2,
    stratify=df['label'],
    random_state=42
)
```

```python
vectorizer = TfidfVectorizer(
    max_features=3000,
    ngram_range=(1,2)
)

X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)
```

```python
model = MultinomialNB(alpha=0.5)
model.fit(X_train_vec, y_train)
```

```python
y_pred = model.predict(X_test_vec)

print("ACCURACY:", accuracy_score(y_test, y_pred))
print("\nCLASSIFICATION REPORT:\n")
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:\n", cm)
```

## test new email

```python
def predict_email(text):
    text = clean_text(text)
    vec = vectorizer.transform([text])
    pred = model.predict(vec)[0]

    if pred == 1:
        print("SPAM ❌")
    else:
        print("HAM ✅")

# test examples
predict_email("Congratulations! You won a free iPhone")
predict_email("Hi, are we meeting tomorrow?")
```
