Features Explained

1.  age - Patient age

2.  sex - Male (1) or Female (0)

3.  cp - Chest pain type (4 types)

4.  trestbps - Resting blood pressure

5.  chol - Cholesterol level

6.  fbs - Fasting blood sugar > 120 mg/dl

7.  restecg - Resting ECG results

8.  thalach - Maximum heart rate achieved

9.  exang - Exercise induced angina

10.  oldpeak - ST depression

11.  slope - Slope of peak exercise ST segment

12.  ca - Number of major vessels colored by fluoroscopy

13.  thal - Thalassemia (blood disorder)

14.  target - 0 or 1 (what you predict)

## Feature engineering

```python
def create_advanced_medical_features(df):
    """Create clinically meaningful features based on medical research"""

    
    # ============================================
    # CARDIOVASCULAR RISK INDICATORS
    # ============================================
    
    # Age-adjusted maximum heart rate (220 - age formula)
    X['max_hr_formula'] = 220 - X['age']
    X['hr_percentage'] = (X['thalach'] / X['max_hr_formula']) * 100
    X['hr_deficit'] = X['max_hr_formula'] - X['thalach']
    
    # Heart rate reserve (important indicator)
    X['hr_reserve'] = X['thalach'] - X['trestbps']
    
    # Exercise capacity (strong predictor)
    X['exercise_capacity'] = X['thalach'] * (1 - X['exang'])
    X['exercise_index'] = X['thalach'] / (X['oldpeak'] + 1)
    
    # ============================================
    # CHOLESTEROL FEATURES
    # ============================================
    
    # Age-adjusted cholesterol
    X['chol_age_ratio'] = X['chol'] / (X['age'] + 1)
    X['high_cholesterol'] = (X['chol'] > 240).astype(int)
    X['borderline_chol'] = ((X['chol'] >= 200) & (X['chol'] <= 240)).astype(int)
    
    # Cholesterol categories (based on medical guidelines)
    X['chol_category'] = pd.cut(X['chol'], 
                                 bins=[0, 200, 240, 300, 600],
                                 labels=[0, 1, 2, 3])
    
    # ============================================
    # BLOOD PRESSURE FEATURES
    # ============================================
    
    # Hypertension stages
    X['hypertension'] = (X['trestbps'] > 140).astype(int)
    X['prehypertension'] = ((X['trestbps'] >= 120) & (X['trestbps'] < 140)).astype(int)
    
    X['bp_category'] = pd.cut(X['trestbps'],
                               bins=[0, 120, 130, 140, 160, 200],
                               labels=[0, 1, 2, 3, 4])
    
    # BP and heart rate interaction
    X['bp_hr_product'] = X['trestbps'] * X['thalach'] / 10000
    
    # ============================================
    # AGE RISK GROUPS
    # ============================================
    
    X['age_group'] = pd.cut(X['age'],
                            bins=[0, 40, 50, 55, 60, 65, 100],
                            labels=[0, 1, 2, 3, 4, 5])
    
    X['senior'] = (X['age'] > 60).astype(int)
    X['middle_age'] = ((X['age'] >= 45) & (X['age'] <= 60)).astype(int)
    
    # ============================================
    # COMPOSITE RISK SCORES
    # ============================================
    
    # Framingham-like risk score
    X['risk_score'] = (
        (X['age'] > 55).astype(int) * 2 +
        (X['chol'] > 240).astype(int) * 2 +
        (X['trestbps'] > 140).astype(int) * 1 +
        X['cp'] +
        X['exang'] * 2 +
        (X['oldpeak'] > 1).astype(int) * 2
    )
    
    # Triple threat (age + cholesterol + BP)
    X['triple_threat'] = (
        (X['age'] > 60) & 
        (X['chol'] > 240) & 
        (X['trestbps'] > 140)
    ).astype(int)
    
    # High risk combo
    X['high_risk_combo'] = (
        (X['cp'] > 0) & 
        (X['exang'] == 1) & 
        (X['oldpeak'] > 1)
    ).astype(int)
    
    # ============================================
    # INTERACTION TERMS
    # ============================================
    
    # Sex-specific features (men at higher risk)
    X['male_age_interaction'] = X['sex'] * X['age']
    X['male_chol_interaction'] = X['sex'] * X['chol']
    
    # Chest pain interactions
    X['cp_sex_interaction'] = X['cp'] * X['sex']
    X['cp_age_interaction'] = X['cp'] * X['age'] / 10
    
    # ST depression interactions
    X['oldpeak_exang'] = X['oldpeak'] * X['exang']
    X['oldpeak_slope'] = X['oldpeak'] * X['slope']
    
    # Vessels and thal interactions
    X['ca_thal'] = X['ca'] * X['thal']
    
    # ============================================
    # POLYNOMIAL FEATURES (selective)
    # ============================================
    
    X['age_squared'] = X['age'] ** 2
    X['chol_squared'] = X['chol'] ** 2
    X['thalach_squared'] = X['thalach'] ** 2
    
    return X

```

## RNN

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, LSTM, GRU, Dense, Dropout, Bidirectional
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
tf.__version__
```

```python
X_train_rnn = X_train_scaled.reshape(X_train_scaled.shape[0], X_train_scaled.shape[1], 1)
X_test_rnn = X_test_scaled.reshape(X_test_scaled.shape[0], X_test_scaled.shape[1], 1)

print(f"Train shape: {X_train_rnn.shape}")
print(f"Test shape:  {X_test_rnn.shape}")
```

```python
model_rnn = Sequential([
   SimpleRNN(64, activation='tanh', return_sequences=True, 
           input_shape=(X_train_rnn.shape[1], 1)),
 Dropout(0.3),
 SimpleRNN(32, activation='tanh'),
 Dropout(0.3),
 Dense(16, activation='relu'),
 Dense(1, activation='sigmoid')
])
```

```python
model_rnn.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

early_stop = EarlyStopping(monitor='val_loss', patience=20, restore_best_weights=True)
reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=10)

history_rnn = model_rnn.fit(
    X_train_rnn, y_train,
    validation_split=0.2,
    epochs=100,
    batch_size=16,
    callbacks=[early_stop, reduce_lr],
    verbose=1
)
```

```python
y_pred_rnn_prob = model_rnn.predict(X_test_rnn)
y_pred_rnn = (y_pred_rnn_prob > 0.5).astype(int).flatten()

acc_rnn = accuracy_score(y_test, y_pred_rnn)
print(f"\n✅ Simple RNN Accuracy: {acc_rnn:.4f} ({acc_rnn*100:.2f}%)")
print(classification_report(y_test, y_pred_rnn))
```

## Bidirectional LSTM (best RNN variant)

```python
model_bilstm = Sequential([
    Bidirectional(LSTM(64, activation='tanh', return_sequences=True), 
                  input_shape=(X_train_rnn.shape[1], 1)),
    Dropout(0.3),
    Bidirectional(LSTM(32, activation='tanh')),
    Dropout(0.3),
    Dense(16, activation='relu'),
    Dense(1, activation='sigmoid')
])
```

```python
y_pred_bilstm_prob = model_bilstm.predict(X_test_rnn)
y_pred_bilstm = (y_pred_bilstm_prob > 0.5).astype(int).flatten()

acc_bilstm = accuracy_score(y_test, y_pred_bilstm)
print(f"\n✅ BiLSTM Accuracy: {acc_bilstm:.4f} ({acc_bilstm*100:.2f}%)")
print(classification_report(y_test, y_pred_bilstm))
```
