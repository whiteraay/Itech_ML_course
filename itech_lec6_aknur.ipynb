1. Series — негізгі тапсырмалар
temperatures деген Series жасаңыз. Ішінде қала температуралары болсын:

Алматы: 5

Астана: –3

Шымкент: 7

 Осы Series-ті экранға шығарыңыз.
Алматының температурасын индекстің көмегімен алыңыз.
Температурасы 0-ден жоғары қалаларды сүзіңіз.


```python
import pandas as pd
import numpy as np
```

```python
temperatures = pd.Series([5, -3, 7], index=['Алматы', 'Астана', 'Шымкент'])
print(temperatures,'\n')
print(temperatures['Алматы'],'\n')
print(temperatures[temperatures > 0],'\n')
```

2. DataFrame — кестемен жұмыс
Студенттер туралы DataFrame жасаңыз:

name    age score

Aigerim 20  88

Dias    22  92

Miras   21  75

Dana    19  95


Жасы 20-дан жоғары студенттерді таңдаңыз.
Бағандарды тек name және score етіп шығарыңыз.
Ең жоғары балл жинаған студенттің атын табыңыз.




```python
students_data = {
    'name': ['Aigerim', 'Dias', 'Miras', 'Dana'],
    'age': [20, 22, 21, 19],
    'score': [88, 92, 75, 95]
}

students_df = pd.DataFrame(students_data)
print(students_df[students_df['age'] > 20])
```

```python
print(students_df[['name', 'score']])
```

```python
highest_score_student = students_df.loc[students_df['score'].idxmax(), 'name']
print(highest_score_student)
```

4. Шарт бойынша сүзу
Дәрілер туралы DataFrame жасаңыз:

drug price amount

A    1200   5

B    800    10

C    600    3

D    2000   8


Тапсырмалар:
Бағасы 1000 теңгеден қымбат дәрілерді шығарыңыз.


Бағасы 600-800 аралығындағы дәрілерді таңдаңыз.


Қордағы саны (amount) 5-тен аз дәрілерді шығарыңыз.


```python
drugs_data = {
    'drug': ['A', 'B', 'C', 'D'],
    'price': [1200, 800, 600, 2000],
    'amount': [5, 10, 3, 8]
}

drugs_df = pd.DataFrame(drugs_data)
print(drugs_df[drugs_df['price'] > 1000])
print()
print(drugs_df[(drugs_df['price'] >= 600) & (drugs_df['price'] <= 800)])
print()
print(drugs_df[drugs_df['amount'] < 5])
```

5.Тапсырмалар:
Әр бөлім бойынша орташа жалақыны есептеңіз.


Әр бөлімдегі қызметкерлер санын шығарыңыз.


Ең жоғары жалақысы бар бөлімді анықтаңыз.


```python
employees_data = {
    'name': ['Ayan', 'Dana', 'Miras', 'Aida', 'Timur'],
    'department': ['IT', 'HR', 'IT', 'HR', 'Sales'],
    'salary': [400000, 300000, 450000, 320000, 280000]
}

employees_df = pd.DataFrame(employees_data)
print(employees_df.groupby('department')['salary'].mean())
print()
print(employees_df.groupby('department')['name'].count())
print()
```

```python
highest_salary_dept = employees_df.groupby('department')['salary'].mean().idxmax()

print(highest_salary_dept)
```

6. Агрегаттау (sum, mean, max, min)

Берілген кестеге сүйеніп:

Әр бөлім үшін жалақының ең кішісі мен ең жоғарғысын табыңыз.


Әр топтағы жалақының суммасын есептеңіз.


```python
print(employees_df.groupby('department')['salary'].agg(['min', 'max']))
print()
print(employees_df.groupby('department')['salary'].sum())
```

7.Тапсырмалар:

Екі кестені id бойынша біріктіріңіз (merge).

Индексті id етіп қойып, join() қолданып біріктіріңіз.

concat() қолданып students кестесін екі рет артынан қосыңыз.

concat(axis=1) қолданып students және scores кестесін бүйірінен біріктіріңіз.


```python
students = pd.DataFrame({'id': [1, 2, 3], 'name': ['Aigerim', 'Dias', 'Ayan']})
scores = pd.DataFrame({'id': [1, 2, 3], 'math': [88, 75, 95], 'physics': [90, 80, 92]})

merged_df = pd.merge(students, scores, on='id')
print(merged_df)
```

```python
students_indexed = students.set_index('id')
scores_indexed = scores.set_index('id')

joined_df = students_indexed.join(scores_indexed)
print(joined_df)
```

```python
concatenated_rows_df = pd.concat([students, students])
print(concatenated_rows_df)
```

```python
concatenated_cols_df = pd.concat([students, scores.drop('id', axis=1)], axis=1)
print(concatenated_cols_df)
```

8. Қиынырақ деңгей:
Төмендегі DataFrame бойынша әр пәннен жалпы статистика (min, max, mean, median) жасаңыз:



Әр студенттің орташа балын есептеп, жаңа avg бағанын қосыңыз.

Орташа балы 85-тен жоғары студенттерді таңдаңыз.


```python
stats_data = {
    'student': ['Aigerim', 'Dias', 'Ayan'],
    'math': [88, 75, 95],
    'physics': [90, 80, 92],
    'history': [78, 85, 90]
}
```

```python
stats_df = pd.DataFrame(stats_data)
stats_df_indexed = stats_df.set_index('student')
```

```python
general_stats = stats_df_indexed.agg(['min', 'max', 'mean', 'median'])
print(general_stats)
```

```python
stats_df['avg'] = stats_df_indexed.mean(axis=1).values
print(stats_df)
```

```python
high_avg_students = stats_df[stats_df['avg'] > 85]
print(high_avg_students)
```
